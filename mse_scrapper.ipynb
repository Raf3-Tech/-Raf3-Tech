## Develop Data Scraper

### Subtask:
Create a Python script using libraries like `requests` and `BeautifulSoup` to fetch the webpage content and extract the table data. The script should parse rows and columns to gather the relevant market information.


**Reasoning**:
The first step is to import the necessary libraries (`requests` for making HTTP requests and `BeautifulSoup` for parsing HTML) and define the target URL. Then, I will send a GET request to the URL to fetch the webpage content.



```python
import requests
from bs4 import BeautifulSoup

# 2. Define the URL for the market data
url = 'https://mse.co.mw/market/mainboard'

# 3. Send an HTTP GET request to the URL
response = requests.get(url)

# Check if the request was successful
response.raise_for_status() # Raise an HTTPError for bad responses (4xx or 5xx)

print("Webpage content fetched successfully.")
```

**Reasoning**:
Now that the webpage content has been fetched, the next step is to parse the HTML content using BeautifulSoup to make it navigable for data extraction.



```python
soup = BeautifulSoup(response.content, 'html.parser')
print("HTML content parsed successfully.")
```

**Reasoning**:
Now that the HTML content is parsed, I need to locate the main market data table within the `soup` object. I will search for the `table` tag and inspect its content to identify the correct table containing the market data.



```python
market_table = soup.find('table', class_='market-table') # Assuming 'market-table' is the class based on common web structures

if market_table:
    print("Market table found successfully.")
    # Optionally, print a snippet of the table to verify
    # print(market_table.prettify()[:500])
else:
    print("Could not find the market data table. Please re-check the website structure.")
```

```python
print(soup.prettify()[:2000]) # Print a larger snippet to find the table
```

**Reasoning**:
The previous snippet of HTML was not large enough to reveal the market data table. I will now try to find all `<table>` elements in the parsed HTML and print their attributes. This will help identify the correct table and its distinguishing features (like `class` or `id`) for more precise selection.



```python
all_tables = soup.find_all('table')

if all_tables:
    print(f"Found {len(all_tables)} table(s).")
    for i, table in enumerate(all_tables):
        print(f"\n--- Table {i+1} ---")
        if table.attrs:
            print("Attributes:", table.attrs)
        else:
            print("No attributes")
        # Print a small snippet of the table content to help identify it
        print("Snippet:", table.prettify()[:500])
else:
    print("No tables found on the page.")
```

**Reasoning**:
The previous output successfully identified the market data table and its class attributes: `table`, `table-sm`, `table-striped`, `table-borderless`. Now I will use these attributes to accurately locate the table and then extract its headers as per the subtask instructions.



```python
market_table = soup.find('table', class_=['table', 'table-sm', 'table-striped', 'table-borderless'])

if market_table:
    print("Market table found successfully using identified classes.")
    # 6. Extract the table headers
    headers = [th.get_text(strip=True) for th in market_table.find('thead').find_all('th')]
    print("Extracted headers:", headers)
else:
    print("Could not find the market data table with the specified classes. Please re-check the website structure.")
```

**Reasoning**:
Now that the market table and its headers have been successfully identified, the next step is to extract the actual data rows. I will iterate through the `tbody` element of the `market_table` to find all table rows (`<tr>`) and then extract the text content from each data cell (`<td>`) within those rows, storing them in a list of lists.



```python
data = []

if market_table:
    # 7. Iterate through each row (<tr> elements) of the table body
    for row in market_table.find('tbody').find_all('tr'):
        # 8. For each row, extract the data from all its cells (<td> elements)
        row_data = [td.get_text(strip=True) for td in row.find_all('td')]
        data.append(row_data)

    print("Extracted data rows successfully.")
    # Optionally, print the first few rows to verify
    print("First 5 extracted rows:", data[:5])

else:
    print("Market table not found, cannot extract data rows.")
```

**Reasoning**:
Now that both the headers and the data rows have been extracted, the next logical step is to combine them into a structured format suitable for further processing and eventual saving to a CSV file. A pandas DataFrame is an ideal data structure for this purpose.



```python
import pandas as pd

# 9. Store the extracted headers and row data in a pandas DataFrame
df = pd.DataFrame(data, columns=headers)

print("DataFrame created successfully.")
# Display the first few rows of the DataFrame to verify
print(df.head())
```

## Set Up Automated Scheduling

### Subtask:
Implement a scheduling mechanism within the Python script to run the data scraping and CSV export at the specified intervals: Pre-Open (09:00-09:30), Open (09:30-14:30), Close (14:30-15:00), and Post Close (15:00-17:00). This will involve using a suitable Python scheduling library.

**Reasoning**:
Now that the necessary libraries are installed and imported, I will encapsulate the entire scraping and CSV export logic into a single Python function. This function will perform all the steps from fetching the webpage to saving the data to a CSV file with a timestamped filename, as per the subtask instructions.



```python
def scrape_and_save_market_data():
    print(f"[{datetime.now().strftime('%Y-%m-%d %H:%M:%S')}] Starting data scraping...")

    # 1. Define the URL for the market data
    url = 'https://mse.co.mw/market/mainboard'

    try:
        # 2. Send an HTTP GET request to the URL
        response = requests.get(url)
        response.raise_for_status() # Raise an HTTPError for bad responses (4xx or 5xx)

        # 3. Parse the HTML content
        soup = BeautifulSoup(response.content, 'html.parser')

        # 4. Locate the market data table
        market_table = soup.find('table', class_=['table', 'table-sm', 'table-striped', 'table-borderless'])

        if not market_table:
            print("Could not find the market data table. Skipping save.")
            return

        # 5. Extract the table headers
        headers = [th.get_text(strip=True) for th in market_table.find('thead').find_all('th')]

        # 6. Extract data rows
        data = []
        for row in market_table.find('tbody').find_all('tr'):
            row_data = [td.get_text(strip=True) for td in row.find_all('td')]
            data.append(row_data)

        # 7. Create a pandas DataFrame
        df = pd.DataFrame(data, columns=headers)

        # 8. Generate a timestamp-based filename and save to CSV
        current_time = datetime.now()
        timestamp_str = current_time.strftime("%Y%m%d_%H%M%S")
        filename = f"market_data_{timestamp_str}.csv"
        df.to_csv(filename, index=False)

        print(f"[{datetime.now().strftime('%Y-%m-%d %H:%M:%S')}] Market data successfully saved to {filename}")

    except requests.exceptions.RequestException as e:
        print(f"[{datetime.now().strftime('%Y-%m-%d %H:%M:%S')}] HTTP Request failed: {e}")
    except Exception as e:
        print(f"[{datetime.now().strftime('%Y-%m-%d %H:%M:%S')}] An unexpected error occurred: {e}")

print("Scraping and saving logic encapsulated in 'scrape_and_save_market_data' function.")
```

**Reasoning**:
Now that the scraping and saving logic is encapsulated in a function, I will use the `schedule` library to schedule this function to run daily at the specified times (09:00, 09:30, 14:30, and 15:00). After scheduling, I will create an infinite loop with a `time.sleep` delay to continuously check and run pending jobs.



```python
schedule.every().day.at("09:00").do(scrape_and_save_market_data)
schedule.every().day.at("09:30").do(scrape_and_save_market_data)
schedule.every().day.at("14:30").do(scrape_and_save_market_data)
schedule.every().day.at("15:00").do(scrape_and_save_market_data)

print("Scheduling jobs...")

while True:
    schedule.run_pending()
    time.sleep(1) # wait one second
```
